```
import pandas as pd
import os

# Base folder where your SMAP CSVs live
base = "/content/drive/MyDrive"

# List of yearly CSVs
smap_files = {
    2016: os.path.join(base, "SMAP_9km_pixels_2016.csv"),
    2017: os.path.join(base, "SMAP_9km_pixels_2017.csv"),
    2018: os.path.join(base, "SMAP_9km_pixels_2018.csv"),
    2019: os.path.join(base, "SMAP_9km_pixels_2019.csv"),
    2020: os.path.join(base, "SMAP_9km_pixels_2020.csv"),
    # later you can extend this dict with 2021–2024
}

# Folder to save preprocessed outputs
out_base = os.path.join(base, "SMAP_preprocessed")
os.makedirs(out_base, exist_ok=True)

def preprocess_smap_year(year, path):
    """
    Load one SMAP CSV, clean it, and save as Parquet.
    Returns the number of rows processed.
    """
    print(f"\n=== Preprocessing {year} from {path} ===")

    # Read CSV
    df = pd.read_csv(path)
    print("  Raw shape:", df.shape)

    # Drop columns you don't really need
    drop_cols = [c for c in ['system:index', '.geo'] if c in df.columns]
    if drop_cols:
        df = df.drop(columns=drop_cols)

    # Parse dates
    df['date'] = pd.to_datetime(df['date'])
    df['time_start'] = pd.to_datetime(df['time_start'], unit='ms')

    # Add year & day-of-year
    df['year'] = df['date'].dt.year
    df['doy'] = df['date'].dt.dayofyear

    # Downcast numeric columns to save memory
    for col in ['sm', 'lat', 'lon']:
        if col in df.columns:
            df[col] = df[col].astype('float32')

    # Quick sanity check
    print("  Cleaned columns:", df.columns.tolist())
    print("  Cleaned shape:  ", df.shape)
    print("  SM stats:\n", df['sm'].describe())

    # Save as Parquet for efficient storage/reading
    out_path = os.path.join(out_base, f"SMAP_9km_pixels_{year}.parquet")
    df.to_parquet(out_path, index=False)
    print("  Saved preprocessed file to:", out_path)

    return len(df)
```

```
total_rows = 0
for year, path in smap_files.items():
    n = preprocess_smap_year(year, path)
    total_rows += n

print("\nTotal rows across all preprocessed years:", total_rows)
```

```
import pandas as pd
import glob
import os

out_base = "/content/drive/MyDrive/SMAP_preprocessed"
year_files = sorted(glob.glob(os.path.join(out_base, "SMAP_9km_pixels_20*_filtered.parquet")))

# We'll accumulate all years' data needed for climatology in chunks
clim_list = []

for f in year_files:
    print("Adding to climatology from:", os.path.basename(f))
    df = pd.read_parquet(f)[['lat', 'lon', 'doy', 'sm']]
    clim_list.append(df)

# Stack just (lat, lon, doy, sm) for climatology
clim_df = pd.concat(clim_list, ignore_index=True)

# Mean SM for each (lat, lon, doy) across all years
pixel_clim = (
    clim_df
    .groupby(['lat', 'lon', 'doy'])['sm']
    .mean()
    .reset_index()
    .rename(columns={'sm': 'sm_clim'})
)

print(pixel_clim.head(), pixel_clim.shape)

# Save for reuse
pixel_clim_path = os.path.join(out_base, "SMAP_pixel_climatology_doy.parquet")
pixel_clim.to_parquet(pixel_clim_path, index=False)
print("Saved climatology to:", pixel_clim_path)
```

```
import pandas as pd
import os

out_base = "/content/drive/MyDrive/SMAP_preprocessed"
pixel_clim_path = os.path.join(out_base, "SMAP_pixel_climatology_doy.parquet")

pixel_clim = pd.read_parquet(pixel_clim_path)
print(pixel_clim.head())
print(pixel_clim.dtypes)

# Make sure dtypes are compact to save RAM
pixel_clim['lat'] = pixel_clim['lat'].astype('float32')
pixel_clim['lon'] = pixel_clim['lon'].astype('float32')
pixel_clim['doy'] = pixel_clim['doy'].astype('int16')
pixel_clim['sm_clim'] = pixel_clim['sm_clim'].astype('float32')

print("\nAfter downcast:")
print(pixel_clim.dtypes)
```

```
import glob

year_files = sorted(glob.glob(os.path.join(out_base, "SMAP_9km_pixels_20*_filtered.parquet")))
print("Filtered files:")
for f in year_files:
    print(" ", os.path.basename(f))

anom_daily_list = []   # will store (year, daily_series)

for f in year_files:
    print("\nComputing anomalies for:", os.path.basename(f))
    df = pd.read_parquet(f, columns=['date', 'lat', 'lon', 'doy', 'sm'])

    year = int(df['date'].dt.year.iloc[0])
    print("  Year:", year, "  shape:", df.shape)

    # Ensure dtypes match climatology
    df['lat'] = df['lat'].astype('float32')
    df['lon'] = df['lon'].astype('float32')
    df['doy'] = df['doy'].astype('int16')

    # Merge climatology onto this year's data
    df = df.merge(
        pixel_clim,
        on=['lat', 'lon', 'doy'],
        how='left'
    )

    # Anomaly = actual - climatology
    df['sm_anom'] = df['sm'] - df['sm_clim']

    # Daily mean anomaly across CONUS
    daily_anom = df.groupby('date')['sm_anom'].mean().sort_index()
    anom_daily_list.append((year, daily_anom))

    print("  Daily anomaly series length:", len(daily_anom))

    # Free memory
    del df
```

```
import pandas as pd
import os

out_base = "/content/drive/MyDrive/SMAP_preprocessed"

# You already loaded this, but just to be explicit:
county = pd.read_parquet(os.path.join(out_base, "SMAP_county_daily_2016_2020.parquet"))
county['date'] = pd.to_datetime(county['date'])

print(county.head())
print("Shape:", county.shape)

# Sort for sanity
county = county.sort_values(['county_fips', 'date']).reset_index(drop=True)

# Per-county mean and std of sm_mean (2016–2020 climatology)
county['sm_mean_clim'] = county.groupby('county_fips')['sm_mean'].transform('mean')
county['sm_mean_std']  = county.groupby('county_fips')['sm_mean'].transform('std')

# Anomaly and z-score
county['sm_anom'] = county['sm_mean'] - county['sm_mean_clim']
county['sm_z']    = county['sm_anom'] / county['sm_mean_std']

# Simple early-warning trigger: z <= -1 on at least 3 of last 5 days
def compute_trigger(group, z_thresh=-1.0, window=5, min_days=3):
    z = group['sm_z']
    dry_days = (z <= z_thresh).astype(int)
    count_dry = dry_days.rolling(window=window, min_periods=1).sum()
    trigger = (count_dry >= min_days).astype(int)
    return trigger

county['trigger'] = (
    county.groupby('county_fips', group_keys=False)
          .apply(compute_trigger)
)

print(county[['county_fips', 'date', 'sm_mean', 'sm_anom', 'sm_z', 'trigger']].head())

# Save enriched version so we don't have to recompute later
county_with_anom_path = os.path.join(out_base, "SMAP_county_daily_2016_2020_with_anom.parquet")
county.to_parquet(county_with_anom_path, index=False)
print("Saved:", county_with_anom_path)
```

```
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import os

# ---------------------------------------------------------------------
# Paths
# ---------------------------------------------------------------------
out_base = "/content/drive/MyDrive/SMAP_preprocessed"
county_path = os.path.join(out_base, "SMAP_county_daily_2016_2020_with_anom.parquet")

# ---------------------------------------------------------------------
# 1. Load county–daily SMAP with anomalies
# ---------------------------------------------------------------------
county = pd.read_parquet(county_path)
county['date'] = pd.to_datetime(county['date'])

# ---------------------------------------------------------------------
# 2. Compute daily CONUS-mean z-score and find driest/wettest days
# ---------------------------------------------------------------------
daily_z = (
    county.groupby('date')['sm_z']
          .mean()
          .sort_index()
)

driest_date = daily_z.idxmin()
wettest_date = daily_z.idxmax()

print("Driest day (lowest CONUS-mean z):", driest_date.date(), "z =", daily_z.loc[driest_date])
print("Wettest day (highest CONUS-mean z):", wettest_date.date(), "z =", daily_z.loc[wettest_date])

# ---------------------------------------------------------------------
# 3. Load counties shapefile (CONUS) once
# ---------------------------------------------------------------------
counties = gpd.read_file(
    "https://www2.census.gov/geo/tiger/GENZ2020/shp/cb_2020_us_county_5m.zip"
)
counties = counties[['GEOID', 'STATEFP', 'NAME', 'geometry']]

# Optional: CONUS only (drop AK=02, HI=15, PR=72)
conus_states = {str(s).zfill(2) for s in range(1, 57)} - {"02", "15", "72"}
counties = counties[counties['STATEFP'].isin(conus_states)].copy()

# ---------------------------------------------------------------------
# 4. Prepare merged GeoDataFrames for driest and wettest days
# ---------------------------------------------------------------------
def make_gmap_for_date(map_date):
    """Return GeoDataFrame with sm_mean and sm_z for the given date merged onto counties."""
    cd = county[county['date'] == map_date].copy()
    if cd.empty:
        raise ValueError(f"No SMAP county data for date {map_date}")

    gmap = counties.merge(
        cd[['county_fips', 'sm_mean', 'sm_z']],
        left_on='GEOID', right_on='county_fips',
        how='left'
    )
    return gpd.GeoDataFrame(gmap, geometry='geometry', crs="EPSG:4326")

gmap_dry = make_gmap_for_date(driest_date)
gmap_wet = make_gmap_for_date(wettest_date)

# ---------------------------------------------------------------------
# 5. Make a 2x2 figure: SM and SM z for driest & wettest days
# ---------------------------------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Row 0: driest day
ax_sm_dry = axes[0, 0]
gmap_dry.plot(
    column='sm_mean',
    ax=ax_sm_dry,
    legend=True,
    linewidth=0.1,
    edgecolor='black',
    missing_kwds={'color': 'lightgrey'}
)
ax_sm_dry.set_title(f"(a) SMAP county-mean SM on driest day\n{driest_date.date()}")
ax_sm_dry.set_axis_off()

ax_z_dry = axes[0, 1]
gmap_dry.plot(
    column='sm_z',
    ax=ax_z_dry,
    legend=True,
    linewidth=0.1,
    edgecolor='black',
    missing_kwds={'color': 'lightgrey'}
)
ax_z_dry.set_title(f"(b) SMAP county-mean SM z-score on driest day\n{driest_date.date()}")
ax_z_dry.set_axis_off()

# Row 1: wettest day
ax_sm_wet = axes[1, 0]
gmap_wet.plot(
    column='sm_mean',
    ax=ax_sm_wet,
    legend=True,
    linewidth=0.1,
    edgecolor='black',
    missing_kwds={'color': 'lightgrey'}
)
ax_sm_wet.set_title(f"(c) SMAP county-mean SM on wettest day\n{wettest_date.date()}")
ax_sm_wet.set_axis_off()

ax_z_wet = axes[1, 1]
gmap_wet.plot(
    column='sm_z',
    ax=ax_z_wet,
    legend=True,
    linewidth=0.1,
    edgecolor='black',
    missing_kwds={'color': 'lightgrey'}
)
ax_z_wet.set_title(f"(d) SMAP county-mean SM z-score on wettest day\n{wettest_date.date()}")
ax_z_wet.set_axis_off()

plt.tight_layout()

# ---------------------------------------------------------------------
# 6. Save figure
# ---------------------------------------------------------------------
fig_path = os.path.join(out_base, "Figure5_SMAP_driest_and_wettest_days_maps.png")
plt.savefig(fig_path, dpi=300)
plt.close()

print("Saved 2x2 driest/wettest map figure to:", fig_path)
```

```
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import os

# Path to your file (note the double .csv.csv)
usdm_path = "/content/drive/MyDrive/SMAP_preprocessed/USDM/USDM_county_2016_2020.csv.csv"

usdm = pd.read_csv(usdm_path)
usdm.columns = usdm.columns.str.strip()
print("USDM columns:", usdm.columns.tolist())
usdm.head()
```

```
# FIPS → county_fips (5-digit string)
usdm['county_fips'] = usdm['FIPS'].astype(str).str.zfill(5)

# MapDate is YYYYMMDD integer/string → proper datetime
usdm['date'] = pd.to_datetime(usdm['MapDate'].astype(str), format="%Y%m%d")

# Keep relevant columns
usdm = usdm[['county_fips', 'date', 'DSCI']]
print(usdm.head())
```

```
def dsci_to_cat(dsci):
    if dsci >= 500:
        return 'D4'
    elif dsci >= 400:
        return 'D3'
    elif dsci >= 300:
        return 'D2'
    elif dsci >= 200:
        return 'D1'
    elif dsci >= 100:
        return 'D0'
    else:
        return 'None'

usdm['usdm_cat'] = usdm['DSCI'].apply(dsci_to_cat)

sev_map = {'None': 0.0, 'D0': 0.5, 'D1': 1.0, 'D2': 2.0, 'D3': 3.0, 'D4': 4.0}
usdm['usdm_sev'] = usdm['usdm_cat'].map(sev_map).fillna(0.0)

print(usdm.head())
```

```
usdm_daily = usdm.copy()
print("USDM daily shape:", usdm_daily.shape)
```

```
out_base = "/content/drive/MyDrive/SMAP_preprocessed"
county_path = os.path.join(out_base, "SMAP_county_daily_2016_2020_with_anom.parquet")

county = pd.read_parquet(county_path)
county['date'] = pd.to_datetime(county['date'])
county['county_fips'] = county['county_fips'].astype(str).str.zfill(5)

print("SMAP counties:", county['county_fips'].nunique())
```

```
usdm_daily = usdm_daily[usdm_daily['county_fips'].isin(county['county_fips'].unique())]

county_usdm = county.merge(usdm_daily, on=['county_fips', 'date'], how='inner')
print(county_usdm.head())
print("Merged rows:", county_usdm.shape)
```

```
county_usdm['event_d1plus'] = (county_usdm['usdm_sev'] >= 1.0).astype(int)
county_usdm[['county_fips', 'date', 'sm_z', 'trigger', 'DSCI', 'usdm_cat', 'usdm_sev', 'event_d1plus']].head()
```

```
def add_future_event(group, event_col='event_d1plus', L=30):
    group = group.sort_values('date').reset_index(drop=True)
    y = group[event_col].values
    n = len(y)
    fut = np.zeros(n, dtype=int)
    for i in range(n):
        j_end = min(n, i + L + 1)
        fut[i] = int(y[i:j_end].any())
    group[f'{event_col}_future_{L}'] = fut
    return group

lead_times = [7, 14, 21, 30, 45, 60]
for L in lead_times:
    county_usdm = (county_usdm
                   .groupby('county_fips', group_keys=False)
                   .apply(lambda g, L=L: add_future_event(g, 'event_d1plus', L=L)))
```

```
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss

results = []
for L in lead_times:
    label_col = f'event_d1plus_future_{L}'
    mask = county_usdm[label_col].notna()

    y_true = county_usdm.loc[mask, label_col].values
    y_score = county_usdm.loc[mask, 'trigger'].astype(float).values

    if y_true.min() == y_true.max():
        print(f"Lead {L}: only one class present, skipping.")
        continue

    roc = roc_auc_score(y_true, y_score)
    pr = average_precision_score(y_true, y_score)
    brier = brier_score_loss(y_true, y_score)

    results.append({'lead_days': L, 'ROC_AUC': roc, 'PR_AUC': pr, 'Brier': brier})

skill_df = pd.DataFrame(results)
print(skill_df)
```

```
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(skill_df['lead_days'], skill_df['ROC_AUC'], marker='o', label='ROC-AUC')
ax.plot(skill_df['lead_days'], skill_df['PR_AUC'], marker='s', label='PR-AUC')
ax.set_xlabel("Lead time (days)")
ax.set_ylabel("Score")
ax.set_ylim(0, 1)
ax.set_title("SMAP dryness trigger skill vs USDM D1+ events")
ax.legend()

plt.tight_layout()
fig_path = os.path.join(out_base, "Figure6_skill_SMAP_trigger_vs_USDM_DSCI.png")
plt.savefig(fig_path, dpi=300)
plt.close()
print("Saved skill figure to:", fig_path)
```
